```python
import tensorflow as tf
import numpy as np
import os
import ast
```

```python
classes=sorted(os.listdir(r'Validation'))
```

```python
model = tf.keras.models.load_model('characters_model2.keras')

# Path to the image
img_path = r''

# Load the image in grayscale mode and resize it
img = tf.keras.preprocessing.image.load_img(img_path, color_mode='grayscale', target_size=(32, 32))

# Convert the image to a numpy array
img_array = tf.keras.preprocessing.image.img_to_array(img)

# Add a channel dimension if needed (grayscale images have shape (32, 32, 1))
img_array = np.expand_dims(img_array, axis=0)

# Normalize the image
img_array = img_array / 255.0

# Make prediction
predictions = model.predict(img_array)

# Get predicted class index
predicted_class_index = np.argmax(predictions, axis=1)[0]

# Load class labels from file
with open('classes.txt', 'r') as f:
    classes = ast.literal_eval(f.read())

# Print predicted class name
print(f"Predicted Class: {classes[predicted_class_index]}")
```
